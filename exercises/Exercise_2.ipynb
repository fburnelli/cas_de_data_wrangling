## Extract Data

```python
import pandas as pd
import os
import sys  

# Profiling
from ydata_profiling import ProfileReport
```

```python
## Extract Data
file_name = 'dirty-loan-data.csv'
data_folder = '../data'
file_path = os.path.join(os.getcwd(), data_folder, file_name)

df = pd.read_csv(file_path, low_memory=False)
```

```python
df.sample(5)
```

```python
df.info()      
```

```python
df.describe()   
```

### Generate The Profile

```python
profile = ProfileReport(df,
                        title='Loan Dataset Profiling Report',
                        minimal=True
                        )
```

### to Iframe

```python
profile.to_notebook_iframe()
```

### Persist as  HTML

```python
profile.to_file(output_file='lc_loan_sample_profiling.html')
```
